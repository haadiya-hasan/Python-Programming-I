```python
#Line Chart
sns.lineplot(x="age", y="fare", data=df)
plt.show()
```

```python
#Relational PLot(relplot)
sns.relplot(x="age", y="fare", hue="parch", col="embarked", data=df)
plt.show()
```

```python
#Histogram
sns.histplot(df["fare"], bins=50, kde=False)
plt.show()
```

```python
#KDE PLot (kernel density estimate)
sns.kdeplot(df["sibsp"], fill=True)
plt.show()
```

```python
#Distribution Plot (Histogram + KDE)
sns.histplot(df["pclass"], bins=50, kde=True)
plt.show()
```

```python
#Bar Chart
sns.barplot(x="alone", y="pclass", data=df)
plt.show()
```

```python
#Boxplot
sns.boxplot(x="parch", y="fare", data=df)
```

```python
#Violin Plot
sns.violinplot(x="survived", y="fare", hue="alone", data=df, split=True)
```
