```python
#break statement we can stop the loop even if the while condition is true:
i = 1
while i < 10:
  print(i)
  if i == 4:
    break
  i += 1
```

```python
#continue statement we can stop the loop even if the while condition is true:
i = 0
while i < 10:
  i += 1
  if i == 5:
    continue
  print(i)
```

```python
cars=["Ford", "Audi", "Range Rover"]
for i in cars:
    if(i=="Audi"):
        continue
    print(i)
```

```python
#continue statement we can stop the current iteration, and continue with the next:
i = 0
while i < 10:
  i += 1
  if i == 5:
    continue
  elif i==8:
    continue
  print(i)
```

```python
# WAP to find sum of first n natural numbers.
n = int(input("enter the last number: "))
i = 1
sum = 0
while (i <= n):
    sum = sum + i
    i = i + 1
print("The sum is: ", sum)
```

```python
# WAP to find sum of squares of n natural numbers.
n = int(input("enter the last number: "))
i = 1
sum = 0
while (i <= n):
    sum = sum + i*i
    i = i + 1
print("The sum is: ", sum)
```

```python
rows = int(input("Enter number of rows: "))

for i in range(rows):
    for j in range(i+1):
        print(j+1, end="")
    print()
```
