```python
#(i) Write a Python program that prints all even numbers between 1 and 50 using a for loop, skipping multiples of
#10 with the continue statement. Include an option to exit the loop early if a number greater than 40 is encountered
#using the break statement.

for i in range(1, 51):
    if i % 2 == 0:       # only even
        if i % 10 == 0:  # skip multiples of 10
            continue
        if i > 40:       # stop if number greater than 40
            break
        print(i)
```

```python
#(ii) Write a program to check for Palindrome, Armstrong & Fibonacci Series.

# Palindrome : A word, phrase, or sequence that reads the same backwards as forwards

words = input("Enter a word/number: ")
word=words.lower()
if word == word[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")

# Armstrong : A number that is equal to the sum of its own digits, each raised to the 
#power of the total number of digits in the number

num = int(input("Enter a number: "))

sum = 0
temp = num
n = len(str(num))

while temp > 0:
    digit = temp % 10
    sum += digit ** n
    temp //= 10

if sum == num:
    print("Armstrong number")
else:
    print("Not an Armstrong number")

# Fibonacci : 
n = int(input("Enter number of terms: "))x`
a, b = 0, 1
for o in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#(iii) Create a Python module that contains a function to calculate the factorial of a number using recursion. 
def factorial(n):
    if n < 0:
        return "Factorial not defined for negative numbers"
    elif n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

# Take input and display result
num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```
