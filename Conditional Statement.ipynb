```python
#Write a program that determines the largest of three given numbers using
#if-elif-else.

a=int(input("Enter the first number"))
b=int(input("Enter the second number"))
c=int(input("Enter the third number"))
if (a>=b) and (a>=c):
    print("The largest number is:", a)
elif (b>=a) and (b>=c):
    print("The largest number is:", b)
else:
    print("\nThe largest number is:", c)
```

```python
#Write a program that asks the user for a letter and prints if it's a vowel (a,
#e, i, o, u) or a 

letters = input("Enter a letter: ")
letter=letters.lower()
if letter in ('a', 'e', 'i', 'o', 'u'):
    print("The letter is a vowel")
else:
    print("The letter is a consonant")
```

```python
#Write a program that checks if a year is a leap year. A leap year is
#divisible by 4, but not by 100, unless it is also divisible by 400.

year = int(input("Enter a year:"))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("It is a leap year.")
else:
    print("It is not a leap year.")
```

```python
#A company gives a 10% bonus to employees who have worked for more than 5 years. 
#Write a program that takes an employee's salary and years of service as input and 
#calculates their total pay, including the bonus if applicable

salary = float(input("Enter your salary: "))
years = int(input("Enter your years of service: "))

# Check if eligible for bonus
if years > 5:
    bonus = 0.10 * salary
    total_pay = salary + bonus
    print(f"You are eligible for a bonus of ₹",bonus)
else:
    total_pay = salary
    print("You are not eligible for a bonus")

# Display total pay
print(f"Your total pay is ₹",total_pay)
```

```python
#Write a program that checks if a given string is a palindrome

words = input("Enter a word: ")

word=words.lower()

if word == word[::-1]:
    print("This is Palindrome")
else:
    print("This is not Palindrome")
```

```python
# Develop a program that takes three side lengths of a triangle as input and
#determines if it is equilateral, isosceles, or scalene.

a = float(input("Enter the first side:"))
b = float(input("Enter the second side:"))
c = float(input("Enter the third side:"))

if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("The triangle is equilateral.")
    elif a == b or b == c or a == c:
        print("The triangle is isosceles.")
    else:
        print("The triangle is scalene.")
else:
    print("The entered sides do not form a valid triangle.")
```
