```python
#Asking user to input the numbers they want the operations to be performed on
n1=int(input("Enter the first number: " ))
n2=int(input("Enter the second number: "))

#Computing the summation
sum=n1+n2
print("The sum is: ", sum)

#Computing the difference
diff=n1-n2
print("The difference is: ", diff)

#Computing the product
prod=n1*n2
print("The product is: ", prod)

#Displaying the quotient
div=n1/n2
print("The quotient is: ", div)

#Displaying the remainder 
mod=n1%n2
print("The remainder is: ", mod)

#The floor divison
floor=n1//n2
print("The floor division is: ", floor)

#Exponential Problem
exp=n1**n2
print("The exponential solution is: ", exp)
```

```python
#Interchanging list items with and without temporary variable

# Original list
lst = [10, 20, 30, 40, 50]
print("Original List:", lst)

#With Temporary Variable
temp = lst[0]
lst[0] = lst[-1]
lst[-1] = temp
print("After Swap (with temporary values):", lst)

#Reset list
lst = [10, 20, 30, 40, 50]

#Without Temporary Variable
lst[0], lst[-1] = lst[-1], lst[0]
print("After Swap (without temp):", lst)
```

```python
# 1. Creating a dictionary
print("Creating a dictionary:")
user_profile = {
    "Name": "Haadiya",
    "Followers": 400,
    "Location": "NOIDA"}

print("Initial dictionary: ", user_profile)


# 2. Adding and Updating elements
print("\nAdding & Updating elements:")

# Add a new key-value pair
user_profile["Profession"] = "Student"

# Update an existing value
user_profile["Followers"] = 500
print("After additions/updates: ", user_profile)


# 3. Removing elements
print("\nRemoving elements:")

# Remove 'location' using pop()
removed_location = user_profile.pop("Location")
print("Removed '{removed_location}' using pop().")

# Remove 'followers' using del
del user_profile["Followers"]
print("Removed 'Followers' using del.")
print("Dictionary after removals: ", user_profile)

# 4. Copying a dictionary
print("\nCopying a dictionary:")
# Use the copy() method to create a new, independent dictionary

profile_copy = user_profile.copy()
profile_copy["Name"] = "Hasan" # Modify the copy

print("Original dictionary is unchanged: ",user_profile)
print("Modified copy: ", profile_copy)

# 5. Creating and Accessing a Nested Dictionary
print("\nCreating a Nested Dictionary:")

# A dictionary where values are other dictionaries
employees = {
    "E101": {
        "Name": "Alice",
        "Role": "Lead Developer",
    },
    "E102": {
        "Name": "Bob",
        "Role": "Project Manager",
    }
}
print("Nested dictionary created: ", employees)

# Accessing an inner value
alices_role = employees["E101"]["Role"]
print("\nAccessed Alice's role: '{alices_role}'")
```
