```python
sort=wineQT.sort_values("density")
plt.plot(wineQT["density"],color="m",linewidth=2)
plt.grid()
plt.title("Density Trend")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(wineQT["pH"], wineQT["density"])
plt.title("pH vs Density")
plt.xlabel("pH")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize=(6,6))
wineQT["quality"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Quality Percentage")
plt.show()
```

```python
#Showing the outliers
plt.figure(figsize=(8,5))
plt.scatter(range(len(wineQT)), wineQT["citric acid"])
plt.title("Scatter Plot Showing Outliers in Citric Acid")
plt.xlabel("Index")
plt.ylabel("Citric cid")
plt.show()
```
